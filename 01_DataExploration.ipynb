```python
import pandas as pd
from pathlib import Path

# Project paths
DATA_DIR = Path("../Dataset")

# Load datasets
accounts = pd.read_csv(DATA_DIR / "accounts.csv")
subscriptions = pd.read_csv(DATA_DIR / "subscriptions.csv")
feature_usage = pd.read_csv(DATA_DIR / "feature_usage.csv")
support_tickets = pd.read_csv(DATA_DIR / "support_tickets.csv")
churn_events = pd.read_csv(DATA_DIR / "churn_events.csv")

datasets = {
    "accounts": accounts,
    "subscriptions": subscriptions,
    "feature_usage": feature_usage,
    "support_tickets": support_tickets,
    "churn_events": churn_events
}

# Basic exploration
for name, df in datasets.items():
    print("\n" + "=" * 60)
    print(f"{name.upper()}")
    print("=" * 60)

    print("\nShape:")
    print(df.shape)

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nFirst 5 rows:")
    print(df.head())

    print("\nData types:")
    print(df.dtypes)

    print("\nMissing values:")
    print(df.isnull().sum())

    print("\nDuplicate rows:")
    print(df.duplicated().sum())
```

```python


print("=" * 70)
print("DATE/TIME DETECTION & CONVERSION")
print("=" * 70)

date_conversion_report = []

for name, df in datasets.items():

    print(f"\n--- {name.upper()} ---")

    for col in df.columns:

        # Only inspect object/string columns
        if df[col].dtype == "object":

            # Try converting the column to datetime
            converted = pd.to_datetime(
                df[col],
                errors="coerce"
            )

            # Calculate how many non-null values were successfully converted
            original_non_null = df[col].notna().sum()

            if original_non_null > 0:
                conversion_rate = (
                    converted.notna().sum() / original_non_null
                ) * 100
            else:
                conversion_rate = 0

            # Convert only if most non-null values look like dates
            if conversion_rate >= 80:

                df[col] = converted

                date_conversion_report.append({
                    "Dataset": name,
                    "Column": col,
                    "Conversion_Rate_%": round(conversion_rate, 2),
                    "New_Dtype": str(df[col].dtype)
                })

                print(
                    f"✓ Converted: {col} "
                    f"({conversion_rate:.1f}% successfully parsed)"
                )


# --------------------------------------------
# Display conversion report
# --------------------------------------------

print("\n" + "=" * 70)
print("DATE CONVERSION REPORT")
print("=" * 70)

if date_conversion_report:
    date_conversion_report = pd.DataFrame(date_conversion_report)
    display(date_conversion_report)
else:
    print("No columns were automatically converted to datetime.")


# --------------------------------------------
# Verify resulting date columns
# --------------------------------------------

print("\n" + "=" * 70)
print("DATE/TIME COLUMNS AFTER CONVERSION")
print("=" * 70)

for name, df in datasets.items():

    date_cols = df.select_dtypes(
        include=["datetime", "datetimetz"]
    ).columns.tolist()

    print(f"\n{name}:")
    
    if date_cols:
        for col in date_cols:
            print(
                f"  • {col}: "
                f"{df[col].min()} → {df[col].max()}"
            )
    else:
        print("  No datetime columns found.")


print("\n" + "=" * 70)
print("DATE/TIME PREPARATION COMPLETE")
print("=" * 70)
```

```python


print("=" * 70)
print("KEY & RELATIONSHIP VALIDATION")
print("=" * 70)


# ------------------------------------------------
# 1. Identify ID-like columns
# ------------------------------------------------

print("\n1. ID / KEY COLUMNS")
print("-" * 70)

for name, df in datasets.items():

    id_cols = [
        col for col in df.columns
        if (
            col.lower() == "id"
            or col.lower().endswith("_id")
            or "uuid" in col.lower()
        )
    ]

    print(f"\n{name.upper()}:")
    
    if id_cols:
        for col in id_cols:
            unique_count = df[col].nunique(dropna=True)
            total_count = df[col].notna().sum()

            print(
                f"  • {col}: "
                f"{unique_count:,} unique / "
                f"{total_count:,} non-null"
            )
    else:
        print("  No obvious ID columns detected.")


# ------------------------------------------------
# 2. Check uniqueness of ID columns
# ------------------------------------------------

print("\n\n2. ID UNIQUENESS CHECK")
print("-" * 70)

key_analysis = []

for name, df in datasets.items():

    id_cols = [
        col for col in df.columns
        if (
            col.lower() == "id"
            or col.lower().endswith("_id")
            or "uuid" in col.lower()
        )
    ]

    for col in id_cols:

        non_null = df[col].notna().sum()
        unique = df[col].nunique(dropna=True)

        key_analysis.append({
            "Dataset": name,
            "Column": col,
            "Rows": len(df),
            "Unique_Values": unique,
            "Non_Null": non_null,
            "Is_Unique": unique == non_null
        })

if key_analysis:
    key_analysis = pd.DataFrame(key_analysis)
    display(key_analysis)
else:
    print("No ID columns available for analysis.")


# ------------------------------------------------
# 3. Cross-dataset ID matching
# ------------------------------------------------

print("\n\n3. CROSS-DATASET ID MATCHING")
print("-" * 70)

# Expected relationships based on the SaaS dataset structure.
# We only test relationships when both columns actually exist.

possible_relationships = [
    ("accounts", "account_id", "subscriptions", "account_id"),
    ("accounts", "account_id", "feature_usage", "account_id"),
    ("accounts", "account_id", "support_tickets", "account_id"),
    ("accounts", "account_id", "churn_events", "account_id"),

    ("subscriptions", "subscription_id", "feature_usage", "subscription_id"),
    ("subscriptions", "subscription_id", "support_tickets", "subscription_id"),
    ("subscriptions", "subscription_id", "churn_events", "subscription_id")
]

relationship_results = []

for parent_table, parent_col, child_table, child_col in possible_relationships:

    if (
        parent_table in datasets
        and child_table in datasets
        and parent_col in datasets[parent_table].columns
        and child_col in datasets[child_table].columns
    ):

        parent_ids = set(
            datasets[parent_table][parent_col]
            .dropna()
            .unique()
        )

        child_ids = datasets[child_table][child_col].dropna()

        matched = child_ids.isin(parent_ids).sum()
        total = len(child_ids)

        match_rate = (
            matched / total * 100
            if total > 0 else 0
        )

        relationship_results.append({
            "Parent_Table": parent_table,
            "Parent_Key": parent_col,
            "Child_Table": child_table,
            "Child_Key": child_col,
            "Child_Records": total,
            "Matched_Records": matched,
            "Match_Rate_%": round(match_rate, 2)
        })


if relationship_results:
    relationship_results = pd.DataFrame(relationship_results)
    display(relationship_results)
else:
    print(
        "No predefined relationships could be tested "
        "using the actual column names."
    )


# ------------------------------------------------
# 4. Potential orphan records
# ------------------------------------------------

print("\n\n4. ORPHAN RECORD CHECK")
print("-" * 70)

if relationship_results is not None and len(relationship_results) > 0:

    for _, row in relationship_results.iterrows():

        if row["Match_Rate_%"] < 100:

            print(
                f"⚠ {row['Child_Table']}.{row['Child_Key']} → "
                f"{row['Parent_Table']}.{row['Parent_Key']}: "
                f"{100 - row['Match_Rate_%']:.2f}% unmatched"
            )

        else:

            print(
                f"✓ {row['Child_Table']}.{row['Child_Key']} → "
                f"{row['Parent_Table']}.{row['Parent_Key']}: "
                f"100% matched"
            )


print("\n" + "=" * 70)
print("KEY & RELATIONSHIP VALIDATION COMPLETE")
print("=" * 70)
```

```python

print("=" * 70)
print("DATA CLEANING & STANDARDIZATION")
print("=" * 70)

cleaning_report = []

for name, df in datasets.items():

    print(f"\n--- CLEANING: {name.upper()} ---")

    original_rows = len(df)
    original_columns = len(df.columns)

    # --------------------------------------------
    # 1. Standardize column names
    # --------------------------------------------

    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_", regex=False)
        .str.replace("-", "_", regex=False)
    )

    # --------------------------------------------
    # 2. Remove completely empty rows
    # --------------------------------------------

    empty_rows = df.isna().all(axis=1).sum()

    if empty_rows > 0:
        df.dropna(how="all", inplace=True)

    # --------------------------------------------
    # 3. Remove completely empty columns
    # --------------------------------------------

    empty_columns = df.isna().all(axis=0).sum()

    if empty_columns > 0:
        df.dropna(axis=1, how="all", inplace=True)

    # --------------------------------------------
    # 4. Standardize text fields
    # --------------------------------------------

    text_columns = df.select_dtypes(
        include=["object", "string"]
    ).columns

    for col in text_columns:
        df[col] = df[col].str.strip()

    # --------------------------------------------
    # 5. Convert blank strings to missing values
    # --------------------------------------------

    for col in text_columns:
        df[col] = df[col].replace(
            ["", " ", "NA", "N/A", "na", "null", "NULL"],
            pd.NA
        )

    # --------------------------------------------
    # 6. Remove exact duplicate rows
    # --------------------------------------------

    duplicates_before = df.duplicated().sum()

    if duplicates_before > 0:
        df.drop_duplicates(inplace=True)

    # --------------------------------------------
    # 7. Reset index
    # --------------------------------------------

    df.reset_index(drop=True, inplace=True)

    # --------------------------------------------
    # Cleaning report
    # --------------------------------------------

    cleaning_report.append({
        "Dataset": name,
        "Original_Rows": original_rows,
        "Final_Rows": len(df),
        "Duplicate_Rows_Removed": duplicates_before,
        "Empty_Rows_Removed": empty_rows,
        "Empty_Columns_Removed": empty_columns,
        "Columns": len(df.columns)
    })

    print(
        f"Rows: {original_rows:,} → {len(df):,}"
    )

    print(
        f"Duplicate rows removed: {duplicates_before:,}"
    )

    print(
        f"Completely empty rows removed: {empty_rows:,}"
    )

    print(
        f"Completely empty columns removed: {empty_columns:,}"
    )


# ------------------------------------------------
# Cleaning summary
# ------------------------------------------------

print("\n" + "=" * 70)
print("CLEANING SUMMARY")
print("=" * 70)

cleaning_report = pd.DataFrame(cleaning_report)

display(cleaning_report)


# ------------------------------------------------
# Verify datasets after cleaning
# ------------------------------------------------

print("\n" + "=" * 70)
print("POST-CLEANING DATASET STATUS")
print("=" * 70)

for name, df in datasets.items():

    print(
        f"{name:<20} "
        f"{df.shape[0]:>8,} rows × "
        f"{df.shape[1]:>3} columns"
    )

print("\nData cleaning and standardization complete.")
```

```python


print("=" * 70)
print("BUSINESS VALIDITY & ANOMALY CHECKS")
print("=" * 70)

anomaly_report = []


# ------------------------------------------------
# 1. Negative numeric values
# ------------------------------------------------

print("\n1. NEGATIVE NUMERIC VALUES")
print("-" * 70)

for name, df in datasets.items():

    numeric_cols = df.select_dtypes(include="number").columns

    dataset_has_negative = False

    for col in numeric_cols:

        negative_count = (df[col] < 0).sum()

        if negative_count > 0:

            dataset_has_negative = True

            anomaly_report.append({
                "Dataset": name,
                "Check": "Negative numeric values",
                "Column": col,
                "Issue_Count": negative_count,
                "Issue_%": round(
                    negative_count / len(df) * 100, 2
                )
            })

            print(
                f"⚠ {col}: {negative_count:,} negative values"
            )

    if not dataset_has_negative:
        print("✓ No negative numeric values found.")


# ------------------------------------------------
# 2. Zero values in numeric columns
# ------------------------------------------------

print("\n\n2. ZERO VALUES")
print("-" * 70)

for name, df in datasets.items():

    numeric_cols = df.select_dtypes(include="number").columns

    for col in numeric_cols:

        zero_count = (df[col] == 0).sum()

        if zero_count > 0:

            print(
                f"{name}.{col}: "
                f"{zero_count:,} zero values "
                f"({zero_count / len(df) * 100:.2f}%)"
            )


# ------------------------------------------------
# 3. Date consistency checks
# ------------------------------------------------

print("\n\n3. DATE CONSISTENCY")
print("-" * 70)

for name, df in datasets.items():

    date_cols = df.select_dtypes(
        include=["datetime", "datetimetz"]
    ).columns.tolist()

    if not date_cols:
        continue

    print(f"\n{name.upper()}:")

    for col in date_cols:

        future_dates = (
            df[col] > pd.Timestamp.now()
        ).sum()

        null_dates = df[col].isna().sum()

        print(
            f"  {col}: "
            f"{future_dates:,} future dates, "
            f"{null_dates:,} missing dates"
        )


# ------------------------------------------------
# 4. Check common start/end date relationships
# ------------------------------------------------

print("\n\n4. START / END DATE LOGIC")
print("-" * 70)

for name, df in datasets.items():

    date_cols = df.select_dtypes(
        include=["datetime", "datetimetz"]
    ).columns.tolist()

    start_cols = [
        col for col in date_cols
        if "start" in col.lower()
    ]

    end_cols = [
        col for col in date_cols
        if "end" in col.lower()
    ]

    if start_cols and end_cols:

        for start_col in start_cols:
            for end_col in end_cols:

                comparable = df[[start_col, end_col]].dropna()

                invalid_dates = (
                    comparable[end_col] < comparable[start_col]
                ).sum()

                print(
                    f"{name}: "
                    f"{start_col} → {end_col}: "
                    f"{invalid_dates:,} invalid sequences"
                )


# ------------------------------------------------
# 5. Extreme numeric values
# ------------------------------------------------

print("\n\n5. EXTREME NUMERIC VALUES (IQR CHECK)")
print("-" * 70)

for name, df in datasets.items():

    numeric_cols = df.select_dtypes(include="number").columns

    for col in numeric_cols:

        series = df[col].dropna()

        if len(series) < 5:
            continue

        q1 = series.quantile(0.25)
        q3 = series.quantile(0.75)
        iqr = q3 - q1

        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr

        outlier_count = (
            (series < lower_bound) |
            (series > upper_bound)
        ).sum()

        if outlier_count > 0:

            print(
                f"{name}.{col}: "
                f"{outlier_count:,} potential outliers "
                f"({outlier_count / len(series) * 100:.2f}%)"
            )


# ------------------------------------------------
# 6. Final anomaly report
# ------------------------------------------------

print("\n" + "=" * 70)
print("ANOMALY SUMMARY")
print("=" * 70)

if anomaly_report:

    anomaly_report = pd.DataFrame(anomaly_report)

    display(anomaly_report)

else:

    print(
        "No negative-value anomalies were detected."
    )

print("\n" + "=" * 70)
print("BUSINESS VALIDITY CHECK COMPLETE")
print("=" * 70)

print(
    "\nIMPORTANT: No suspicious records were deleted "
    "based on these checks."
)
```

```python

print("=" * 70)
print("FEATURE & COLUMN PROFILING")
print("=" * 70)

column_profile = []

for name, df in datasets.items():

    for col in df.columns:

        series = df[col]

        # Basic information
        data_type = str(series.dtype)
        missing_count = series.isna().sum()
        missing_pct = round(
            (missing_count / len(df)) * 100, 2
        )

        unique_count = series.nunique(dropna=True)

        # Identify column type
        if pd.api.types.is_numeric_dtype(series):
            column_type = "Numerical"

        elif pd.api.types.is_datetime64_any_dtype(series):
            column_type = "Date/Time"

        else:
            column_type = "Categorical/Text"

        # Example values
        sample_values = (
            series.dropna()
            .astype(str)
            .unique()[:5]
        )

        sample_values = ", ".join(sample_values)

        column_profile.append({
            "Dataset": name,
            "Column": col,
            "Data_Type": data_type,
            "Column_Type": column_type,
            "Unique_Values": unique_count,
            "Missing_Count": missing_count,
            "Missing_%": missing_pct,
            "Example_Values": sample_values
        })


# ------------------------------------------------
# Create profile dataframe
# ------------------------------------------------

column_profile = pd.DataFrame(column_profile)

print("\nComplete column profile:")
display(column_profile)


# ------------------------------------------------
# Numerical feature summary
# ------------------------------------------------

print("\n" + "=" * 70)
print("NUMERICAL FEATURES")
print("=" * 70)

numerical_profile = column_profile[
    column_profile["Column_Type"] == "Numerical"
].copy()

display(numerical_profile)


# ------------------------------------------------
# Categorical feature summary
# ------------------------------------------------

print("\n" + "=" * 70)
print("CATEGORICAL / TEXT FEATURES")
print("=" * 70)

categorical_profile = column_profile[
    column_profile["Column_Type"] == "Categorical/Text"
].copy()

display(categorical_profile)


# ------------------------------------------------
# Date features
# ------------------------------------------------

print("\n" + "=" * 70)
print("DATE / TIME FEATURES")
print("=" * 70)

date_profile = column_profile[
    column_profile["Column_Type"] == "Date/Time"
].copy()

if len(date_profile) > 0:
    display(date_profile)
else:
    print("No datetime columns detected.")


# ------------------------------------------------
# High-cardinality categorical fields
# ------------------------------------------------

print("\n" + "=" * 70)
print("HIGH-CARDINALITY CATEGORICAL FEATURES")
print("=" * 70)

high_cardinality = categorical_profile[
    categorical_profile["Unique_Values"] > 20
].sort_values(
    "Unique_Values",
    ascending=False
)

if len(high_cardinality) > 0:
    display(high_cardinality)
else:
    print("No high-cardinality categorical columns detected.")


print("\n" + "=" * 70)
print("FEATURE PROFILING COMPLETE")
print("=" * 70)
```

```python
# ============================================
# CELL 14 — CUSTOMER-LEVEL DATASET FOUNDATION
# ============================================

print("=" * 70)
print("CUSTOMER-LEVEL ANALYTICAL DATASET")
print("=" * 70)

# ------------------------------------------------
# 1. Identify account IDs in each dataset
# ------------------------------------------------

print("\n1. ACCOUNT ID COLUMNS")
print("-" * 70)

for name, df in datasets.items():

    account_id_cols = [
        col for col in df.columns
        if "account" in col.lower() and "id" in col.lower()
    ]

    print(f"{name:<20}: {account_id_cols}")


# ------------------------------------------------
# 2. Create base account table
# ------------------------------------------------

accounts_df = datasets["accounts"].copy()

account_id_candidates = [
    col for col in accounts_df.columns
    if "account" in col.lower() and "id" in col.lower()
]

if not account_id_candidates:
    raise KeyError(
        "No account ID column found in accounts dataset."
    )

ACCOUNT_ID = account_id_candidates[0]

customer_analytics = (
    accounts_df
    .drop_duplicates(subset=[ACCOUNT_ID])
    .copy()
)

print("\nBase customer table created.")
print(f"Account ID: {ACCOUNT_ID}")
print(f"Unique accounts: {len(customer_analytics):,}")


# ------------------------------------------------
# 3. Subscription-level aggregation
# ------------------------------------------------

subscriptions_df = datasets["subscriptions"].copy()

subscription_account_cols = [
    col for col in subscriptions_df.columns
    if "account" in col.lower() and "id" in col.lower()
]

if subscription_account_cols:

    SUB_ACCOUNT_ID = subscription_account_cols[0]

    subscription_agg = (
        subscriptions_df
        .groupby(SUB_ACCOUNT_ID)
        .agg(
            subscription_count=(
                SUB_ACCOUNT_ID, "count"
            )
        )
        .reset_index()
    )

    customer_analytics = customer_analytics.merge(
        subscription_agg,
        left_on=ACCOUNT_ID,
        right_on=SUB_ACCOUNT_ID,
        how="left"
    )

    if SUB_ACCOUNT_ID != ACCOUNT_ID:
        customer_analytics.drop(
            columns=[SUB_ACCOUNT_ID],
            inplace=True
        )

    print(
        f"\n✓ Subscription aggregation added: "
        f"{len(subscription_agg):,} accounts"
    )

else:

    print(
        "\n⚠ No account ID found in subscriptions."
    )


# ------------------------------------------------
# 4. Feature usage aggregation
# ------------------------------------------------

feature_usage_df = datasets["feature_usage"].copy()

usage_account_cols = [
    col for col in feature_usage_df.columns
    if "account" in col.lower() and "id" in col.lower()
]

if usage_account_cols:

    USAGE_ACCOUNT_ID = usage_account_cols[0]

    usage_agg = (
        feature_usage_df
        .groupby(USAGE_ACCOUNT_ID)
        .agg(
            usage_record_count=(
                USAGE_ACCOUNT_ID, "count"
            )
        )
        .reset_index()
    )

    customer_analytics = customer_analytics.merge(
        usage_agg,
        left_on=ACCOUNT_ID,
        right_on=USAGE_ACCOUNT_ID,
        how="left"
    )

    if USAGE_ACCOUNT_ID != ACCOUNT_ID:
        customer_analytics.drop(
            columns=[USAGE_ACCOUNT_ID],
            inplace=True
        )

    print(
        f"✓ Feature usage aggregation added: "
        f"{len(usage_agg):,} accounts"
    )

else:

    print(
        "\n⚠ No account ID found in feature usage."
    )


# ------------------------------------------------
# 5. Support aggregation
# ------------------------------------------------

support_df = datasets["support_tickets"].copy()

support_account_cols = [
    col for col in support_df.columns
    if "account" in col.lower() and "id" in col.lower()
]

if support_account_cols:

    SUPPORT_ACCOUNT_ID = support_account_cols[0]

    support_agg = (
        support_df
        .groupby(SUPPORT_ACCOUNT_ID)
        .agg(
            support_ticket_count=(
                SUPPORT_ACCOUNT_ID, "count"
            )
        )
        .reset_index()
    )

    customer_analytics = customer_analytics.merge(
        support_agg,
        left_on=ACCOUNT_ID,
        right_on=SUPPORT_ACCOUNT_ID,
        how="left"
    )

    if SUPPORT_ACCOUNT_ID != ACCOUNT_ID:
        customer_analytics.drop(
            columns=[SUPPORT_ACCOUNT_ID],
            inplace=True
        )

    print(
        f"✓ Support aggregation added: "
        f"{len(support_agg):,} accounts"
    )

else:

    print(
        "\n⚠ No account ID found in support tickets."
    )


# ------------------------------------------------
# 6. Churn aggregation
# ------------------------------------------------

churn_df = datasets["churn_events"].copy()

churn_account_cols = [
    col for col in churn_df.columns
    if "account" in col.lower() and "id" in col.lower()
]

if churn_account_cols:

    CHURN_ACCOUNT_ID = churn_account_cols[0]

    churn_agg = (
        churn_df
        .groupby(CHURN_ACCOUNT_ID)
        .agg(
            churn_event_count=(
                CHURN_ACCOUNT_ID, "count"
            )
        )
        .reset_index()
    )

    # Create binary churn flag
    churn_agg["churned_flag"] = 1

    customer_analytics = customer_analytics.merge(
        churn_agg,
        left_on=ACCOUNT_ID,
        right_on=CHURN_ACCOUNT_ID,
        how="left"
    )

    if CHURN_ACCOUNT_ID != ACCOUNT_ID:
        customer_analytics.drop(
            columns=[CHURN_ACCOUNT_ID],
            inplace=True
        )

    print(
        f"✓ Churn aggregation added: "
        f"{len(churn_agg):,} accounts"
    )

else:

    print(
        "\n⚠ No account ID found in churn events."
    )


# ------------------------------------------------
# 7. Fill aggregation-based missing values
# ------------------------------------------------

aggregation_columns = [
    "subscription_count",
    "usage_record_count",
    "support_ticket_count",
    "churn_event_count",
    "churned_flag"
]

for col in aggregation_columns:

    if col in customer_analytics.columns:
        customer_analytics[col] = (
            customer_analytics[col]
            .fillna(0)
        )


# ------------------------------------------------
# 8. Final customer-level dataset
# ------------------------------------------------

print("\n" + "=" * 70)
print("CUSTOMER ANALYTICAL DATASET")
print("=" * 70)

print(
    f"Rows: {customer_analytics.shape[0]:,}"
)

print(
    f"Columns: {customer_analytics.shape[1]:,}"
)

print("\nColumns:")
print(customer_analytics.columns.tolist())

print("\nFirst 10 rows:")
display(customer_analytics.head(10))


# ------------------------------------------------
# 9. Verify one row per account
# ------------------------------------------------

print("\n" + "=" * 70)
print("ACCOUNT GRAIN VALIDATION")
print("=" * 70)

duplicate_accounts = (
    customer_analytics[ACCOUNT_ID]
    .duplicated()
    .sum()
)

print(
    f"Duplicate account IDs: {duplicate_accounts:,}"
)

if duplicate_accounts == 0:
    print("✓ Confirmed: one row per account.")
else:
    print("⚠ Multiple rows per account detected.")


print("\n" + "=" * 70)
print("CUSTOMER-LEVEL FOUNDATION COMPLETE")
print("=" * 70)
```

```python

print("=" * 70)
print("CUSTOMER SEGMENT & CHURN ANALYSIS")
print("=" * 70)

# Candidate business dimensions
segment_columns = [
    "industry",
    "country",
    "plan_tier",
    "is_trial",
    "referral_source"
]

# Keep only columns that actually exist
segment_columns = [
    col for col in segment_columns
    if col in customer_analytics.columns
]

print("\nDimensions analyzed:")
print(segment_columns)


# Analyze each dimension
for segment_col in segment_columns:

    print("\n" + "=" * 70)
    print(f"{segment_col.upper()} VS CHURN")
    print("=" * 70)

    segment_churn = (
        customer_analytics
        .groupby(segment_col, dropna=False)
        .agg(
            customers=("account_id", "nunique"),
            churned_customers=("historical_churn_flag", "sum"),
            avg_mrr=("mrr_amount_sum", "mean"),
            avg_usage=("usage_records", "mean"),
            avg_support_tickets=("support_ticket_count", "mean")
        )
        .reset_index()
    )

    segment_churn["churn_rate_pct"] = (
        segment_churn["churned_customers"]
        / segment_churn["customers"]
        * 100
    ).round(2)

    # Revenue contribution
    segment_churn["estimated_total_mrr"] = (
        segment_churn["customers"]
        * segment_churn["avg_mrr"]
    ).round(2)

    # Sort by customer count so tiny groups don't dominate the table
    segment_churn = segment_churn.sort_values(
        "customers",
        ascending=False
    )

    display(segment_churn)


# Trial customers specifically
if "is_trial" in customer_analytics.columns:

    print("\n" + "=" * 70)
    print("TRIAL VS NON-TRIAL CUSTOMER ANALYSIS")
    print("=" * 70)

    trial_analysis = (
        customer_analytics
        .groupby("is_trial", dropna=False)
        .agg(
            customers=("account_id", "nunique"),
            churned_customers=("historical_churn_flag", "sum"),
            avg_mrr=("mrr_amount_sum", "mean"),
            avg_usage=("usage_records", "mean"),
            avg_features=("distinct_features_used", "mean")
        )
        .reset_index()
    )

    trial_analysis["churn_rate_pct"] = (
        trial_analysis["churned_customers"]
        / trial_analysis["customers"]
        * 100
    ).round(2)

    display(trial_analysis)


print("\n✓ Customer segment vs churn analysis completed.")
```

```python


print("=" * 70)
print("CUSTOMER HEALTH SEGMENTATION")
print("=" * 70)

# Create percentile-based scores so different metrics
# can be compared on a common 0-100 scale.

def percentile_score(series):
    return series.rank(pct=True) * 100


# 1. Engagement score
customer_analytics["engagement_score"] = (
    percentile_score(customer_analytics["usage_records"])
    + percentile_score(customer_analytics["distinct_features_used"])
    + percentile_score(customer_analytics["active_usage_days"])
) / 3


# 2. Support experience score
# Higher satisfaction = better
# Lower resolution time = better
# Lower escalation count = better

satisfaction_score = percentile_score(
    customer_analytics["avg_satisfaction_score"].fillna(
        customer_analytics["avg_satisfaction_score"].median()
    )
)

resolution_score = 100 - percentile_score(
    customer_analytics["avg_resolution_time_hours"].fillna(
        customer_analytics["avg_resolution_time_hours"].median()
    )
)

escalation_score = 100 - percentile_score(
    customer_analytics["total_escalations"]
)

customer_analytics["support_health_score"] = (
    satisfaction_score
    + resolution_score
    + escalation_score
) / 3


# 3. Revenue value score
customer_analytics["revenue_value_score"] = percentile_score(
    customer_analytics["mrr_amount_sum"]
)


# 4. Overall customer health score
customer_analytics["customer_health_score"] = (
    customer_analytics["engagement_score"] * 0.50
    + customer_analytics["support_health_score"] * 0.25
    + customer_analytics["revenue_value_score"] * 0.25
).round(2)


# 5. Health segments
customer_analytics["customer_health_segment"] = pd.cut(
    customer_analytics["customer_health_score"],
    bins=[-1, 25, 50, 75, 101],
    labels=[
        "At Risk",
        "Needs Attention",
        "Healthy",
        "Highly Engaged"
    ]
)


# 6. Segment performance
health_analysis = (
    customer_analytics
    .groupby("customer_health_segment", observed=False)
    .agg(
        customers=("account_id", "nunique"),
        churned_customers=("historical_churn_flag", "sum"),
        avg_health_score=("customer_health_score", "mean"),
        avg_mrr=("mrr_amount_sum", "mean"),
        avg_usage=("usage_records", "mean"),
        avg_features=("distinct_features_used", "mean"),
        avg_support_tickets=("support_ticket_count", "mean"),
        avg_satisfaction=("avg_satisfaction_score", "mean")
    )
    .reset_index()
)

health_analysis["churn_rate_pct"] = (
    health_analysis["churned_customers"]
    / health_analysis["customers"]
    * 100
).round(2)

display(health_analysis)


# 7. Customer-level health view
print("\nCUSTOMER HEALTH SAMPLE")
print("-" * 70)

health_columns = [
    "account_id",
    "plan_tier",
    "mrr_amount_sum",
    "usage_records",
    "distinct_features_used",
    "support_ticket_count",
    "avg_satisfaction_score",
    "engagement_score",
    "support_health_score",
    "revenue_value_score",
    "customer_health_score",
    "customer_health_segment",
    "historical_churn_flag"
]

display(
    customer_analytics[health_columns]
    .sort_values("customer_health_score")
    .head(20)
)

print("\n✓ Customer health segmentation completed.")
```

```python


print("=" * 70)
print("CHURN DRIVER ASSOCIATION ANALYSIS")
print("=" * 70)

# Numeric features available for analysis
numeric_cols = customer_analytics.select_dtypes(
    include="number"
).columns.tolist()

# Remove identifiers and churn-derived variables
excluded_cols = [
    "historical_churn_flag",
    "churn_flag",
    "churn_event_count",
    "reactivation_count",
    "preceding_upgrade_count",
    "preceding_downgrade_count",
    "account_id"
]

analysis_features = [
    col for col in numeric_cols
    if col not in excluded_cols
]

# Calculate correlation with historical churn
churn_correlation = (
    customer_analytics[analysis_features + ["historical_churn_flag"]]
    .corr(numeric_only=True)["historical_churn_flag"]
    .drop("historical_churn_flag")
    .dropna()
    .sort_values()
)

correlation_table = pd.DataFrame({
    "feature": churn_correlation.index,
    "correlation_with_churn": churn_correlation.values
})

correlation_table["absolute_correlation"] = (
    correlation_table["correlation_with_churn"]
    .abs()
)

correlation_table = correlation_table.sort_values(
    "absolute_correlation",
    ascending=False
)

print("\nTop features associated with historical churn:")
display(correlation_table.head(20))


# 1. Strongest positive associations
print("\n" + "=" * 70)
print("STRONGEST POSITIVE ASSOCIATIONS")
print("=" * 70)

positive = correlation_table[
    correlation_table["correlation_with_churn"] > 0
].head(10)

display(positive)


# 2. Strongest negative associations
print("\n" + "=" * 70)
print("STRONGEST NEGATIVE ASSOCIATIONS")
print("=" * 70)

negative = correlation_table[
    correlation_table["correlation_with_churn"] < 0
].sort_values(
    "correlation_with_churn"
).head(10)

display(negative)


# 3. Save shortlist for later ML work
candidate_features = correlation_table.head(15)["feature"].tolist()

print("\nCandidate analytical features for further investigation:")
print(candidate_features)

print("\n✓ Churn driver association analysis completed.")
print("⚠ Correlation shows association, not causation.")
```

```python


print("=" * 70)
print("KEY CHURN DRIVER SEGMENTATION")
print("=" * 70)

# Create interpretable bands for the strongest business variables

# 1. Support ticket volume
customer_analytics["ticket_band"] = pd.cut(
    customer_analytics["support_ticket_count"],
    bins=[-1, 0, 2, 5, float("inf")],
    labels=[
        "No Tickets",
        "1-2 Tickets",
        "3-5 Tickets",
        "6+ Tickets"
    ]
)

# 2. Feature adoption
customer_analytics["feature_adoption_band"] = pd.cut(
    customer_analytics["distinct_features_used"],
    bins=[-1, 2, 5, 10, float("inf")],
    labels=[
        "0-2 Features",
        "3-5 Features",
        "6-10 Features",
        "11+ Features"
    ]
)

# 3. Product usage
customer_analytics["usage_band"] = pd.qcut(
    customer_analytics["usage_usage_count_sum"],
    q=4,
    labels=[
        "Low Usage",
        "Medium-Low Usage",
        "Medium-High Usage",
        "High Usage"
    ],
    duplicates="drop"
)

# 4. Seats
customer_analytics["seat_band"] = pd.qcut(
    customer_analytics["seats_sum"],
    q=4,
    labels=[
        "Small",
        "Medium-Low",
        "Medium-High",
        "Large"
    ],
    duplicates="drop"
)


# Helper function
def churn_summary(data, group_col):

    result = (
        data
        .groupby(group_col, observed=False)
        .agg(
            customers=("account_id", "nunique"),
            churned_customers=("historical_churn_flag", "sum"),
            avg_mrr=("mrr_amount_sum", "mean"),
            avg_usage=("usage_usage_count_sum", "mean"),
            avg_support_tickets=("support_ticket_count", "mean")
        )
        .reset_index()
    )

    result["churn_rate_pct"] = (
        result["churned_customers"]
        / result["customers"]
        * 100
    ).round(2)

    return result


# Analyze each driver
driver_bands = [
    "ticket_band",
    "feature_adoption_band",
    "usage_band",
    "seat_band"
]

for driver in driver_bands:

    print("\n" + "=" * 70)
    print(driver.upper())
    print("=" * 70)

    display(
        churn_summary(
            customer_analytics,
            driver
        )
    )


print("\n✓ Key churn driver segmentation completed.")
```
